```python
import pandas as pd
import matplotlib.pyplot as plt
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))

x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))
m = (y2 - y1) / (x2 - x1)
dx = x2 - x1
dy = y2 - y1
print("\nSlope m =", round(m, 2))
print("Δx =", dx)
print("Δy =", dy)
if abs(dx) >= abs(dy):

    delta_x = 1 if dx > 0 else -1
    delta_y = m * delta_x
else:
    delta_y = 1 if dy > 0 else -1
    delta_x = delta_y / m

print("Assigned Δx =", round(delta_x, 2))
print("Assigned Δy =", round(delta_y, 2))

steps = int(max(abs(dx), abs(dy)))
xi = x1
yi = y1

data = []

for i in range(steps + 1):
    x_next = xi + delta_x
    y_next = yi + delta_y

    if i < steps:
        data.append([
            round(xi, 2),
            round(yi, 2),
            round(x_next, 2),
            round(y_next, 2)
        ])
    xi = x_next
    yi = y_next
table = pd.DataFrame(
    data,
    columns=[
        "xi",
        "yi",
        "xi+1",
        "yi+1"
    ]
)
print("\nDDA Table:")
display(table)
print("\nRound Off Points:")
xi = x1
yi = y1
points = []
for i in range(steps + 1):

    points.append((round(xi), round(yi)))
    xi += delta_x
    yi += delta_y

print(points)
x_plot = [p[0] for p in points]
y_plot = [p[1] for p in points]

plt.figure(figsize=(7, 5))
plt.plot(
    x_plot,
    y_plot,
    marker='o',
    linewidth=2
)
for x, y in points:
    plt.text(x + 0.05, y + 0.05, f"({x},{y})")

plt.grid(True)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("DDA Line Drawing Algorithm")

plt.show()
```
