```python
import matplotlib.pyplot as plt
import pandas as pd
def bresenham(x1, y1, x2, y2):
    m = (y2 - y1) / (x2 - x1)
    dx = x2 - x1
    dy = y2 - y1
    p = 2 * dy - dx
    print("\nBresenham's Line Drawing Algorithm")
    print("Slope (m) =", round(m, 2))
    print("Δx =", dx)
    print("Δy =", dy)
    print("Initial P =", p)
    points = []
    table = []
    x = x1
    y = y1
    points.append((x, y))
    for i in range(dx):
        old_x = x
        old_y = y
        old_p = p
        if p < 0:
            x = x + 1
            y = y
            p = p + 2 * dy
            case = "P < 0"
        else:
            x = x + 1
            y = y + 1
            p = p + 2 * dy - 2 * dx
            case = "P >= 0"
        points.append((x, y))
        table.append([
            i + 1,
            old_x,
            old_y,
            old_p,
            case,
            x,
            y,
            p
        ])
    df = pd.DataFrame(table, columns=[
        "Step",
        "xi",
        "yi",
        "Pi",
        "Case",
        "xi+1",
        "yi+1",
        "Pi+1"
    ])

    print("\nBresenham Calculation Table:")
    display(df)


    print("\nGenerated Points:")
    print(points)

    x_values = [p[0] for p in points]
    y_values = [p[1] for p in points]

    plt.figure(figsize=(8, 6))

    plt.plot(
        x_values,
        y_values,
        marker='o',
        linewidth=2
    )

    for x, y in points:
        plt.text(
            x + 0.08,
            y + 0.08,
            f"({x},{y})"
        )
    plt.grid(True)
    plt.xlabel("X-axis")
    plt.ylabel("Y-axis")
    plt.title("Bresenham's Line Drawing Algorithm")
    plt.xticks(range(min(x_values)-1, max(x_values)+2))
    plt.yticks(range(min(y_values)-1, max(y_values)+2))
    plt.show()
x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))

x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))
bresenham(x1, y1, x2, y2)
```
